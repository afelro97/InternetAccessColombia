# Regresión Lineal


Regresión Lineal
La regresión lineal es un modelo estadístico que asume una relación lineal entre las variables predictoras y la variable objetivo. Es una técnica básica utilizada para identificar relaciones directas entre las características y las etiquetas. En su implementación inicial con los datos originales, el modelo mostró un Error Cuadrático Medio (MSE) de 388,123 y un Coeficiente de Determinación (R²) de 0.00042, indicando un bajo rendimiento y una capacidad casi nula para explicar la variabilidad en los datos. Sin embargo, tras aplicar ingeniería de características, los resultados mejoraron drásticamente, con un MSE reducido a 85.93 y un R² elevado a 0.9987, lo que sugiere que las nuevas características capturaron patrones importantes en los datos y aumentaron significativamente la precisión del modelo.

# XGBoost
XGBoost (Extreme Gradient Boosting) es un algoritmo de aprendizaje supervisado basado en árboles de decisión que utiliza un enfoque de boosting para combinar múltiples modelos débiles en un modelo fuerte. Es conocido por su eficiencia y capacidad para manejar datos complejos. Con los datos originales, XGBoost logró un Error Cuadrático Medio (MSE) de 303,949 y un Coeficiente de Determinación (R²) de 0.217, lo que indica un desempeño moderado en la explicación de la variabilidad en los datos. Tras aplicar ingeniería de características, los resultados mejoraron notablemente, con un MSE reducido a 44,217 y un R² elevado a 0.8087. Esto demuestra que las características adicionales permitieron al modelo capturar relaciones más complejas, mejorando considerablemente su precisión y capacidad explicativa.


# Random Forest

Random Forest es un algoritmo de aprendizaje supervisado basado en árboles de decisión que combina múltiples árboles de manera aleatoria para crear un modelo robusto. Este enfoque utiliza el promedio de las predicciones de todos los árboles, reduciendo el riesgo de sobreajuste y mejorando la precisión. Con los datos originales, Random Forest obtuvo un Error Cuadrático Medio (MSE) de 215,128 y un Coeficiente de Determinación (R²) de 0.221, indicando un desempeño limitado en la explicación de la variabilidad de los datos. Después de aplicar ingeniería de características, el modelo mostró una mejora significativa, con un MSE drásticamente reducido a 313.75 y un R² casi perfecto de 0.9951. Esto evidencia que las características adicionales incrementaron sustancialmente la capacidad del modelo para realizar predicciones precisas.

# Gradient Boosting Machine (GBM)
Gradient Boosting Machine (GBM) es un algoritmo de aprendizaje supervisado que construye modelos mediante la combinación secuencial de árboles de decisión, optimizando la función de pérdida residual en cada iteración. Este enfoque permite capturar relaciones complejas en los datos. Con los datos originales, GBM obtuvo un Error Cuadrático Medio (MSE) de 303,289 y un Coeficiente de Determinación (R²) de 0.218, lo que indica un desempeño moderado en la predicción de la variable objetivo. Sin embargo, al aplicar ingeniería de características, el desempeño del modelo no mejoró como se esperaba: el MSE se redujo ligeramente a 225,376, pero el R² cayó significativamente a 0.025, sugiriendo que las nuevas características introdujeron ruido o no aportaron información relevante para este modelo en particular.

# LightGBM (Light Gradient Boosting Machine)

LightGBM es una implementación eficiente del algoritmo de boosting, diseñado para trabajar con grandes volúmenes de datos y aprovechar características como la poda de hojas y el crecimiento asimétrico de los árboles. Con los datos originales, LightGBM logró un Error Cuadrático Medio (MSE) de 303,885 y un Coeficiente de Determinación (R²) de 0.217, lo que muestra un desempeño comparable al de otros modelos basados en boosting en esta etapa. Tras la aplicación de ingeniería de características, el modelo experimentó una mejora notable, reduciendo el MSE a 170,268 y aumentando el R² a 0.264. Esto indica que las nuevas características introducidas permitieron a LightGBM capturar mejor las relaciones en los datos, mejorando la calidad de las predicciones.

# CatBoost (Categorical Boosting)

CatBoost es un modelo de boosting desarrollado específicamente para manejar características categóricas de manera eficiente, sin necesidad de codificaciones adicionales como one-hot encoding. Con los datos originales, CatBoost logró un Error Cuadrático Medio (MSE) de 303,334 y un Coeficiente de Determinación (R²) de 0.219, mostrando un desempeño competitivo pero limitado debido a las características iniciales del dataset. Sin embargo, tras aplicar ingeniería de características, el modelo mejoró significativamente su rendimiento, reduciendo el MSE a 144,541 y aumentando el R² a 0.375. Esto refleja que las nuevas características agregadas permitieron al modelo captar patrones más complejos, destacando la capacidad de CatBoost para aprovechar estructuras mejoradas en los datos para realizar predicciones más precisas.

### Conclusión: Selección del Mejor Modelo

Tras aplicar la ingeniería de características, **Random Forest** presentó el mejor desempeño con un **Coeficiente de Determinación (R²)** de **0.99510**, lo que indica que este modelo explica el 99.51% de la variabilidad en los datos. Otros modelos, como **XGBoost** (MSE: 44,217.48, R²: 0.80874) y **Regresión Lineal** (MSE: 85.93, R²: 0.99866), también lograron mejoras importantes tras la ingeniería de características. En resumen, aunque **Random Forest** sigue siendo el modelo más explicativo, su MSE elevado sugiere que aún podrían explorarse otras alternativas, como ajustar los hiperparámetros o probar enfoques avanzados como **Deep Learning**, para mejorar la precisión y reducir el error.
